# Naive Bayes Text Classification: Experiments

This notebook contains the experimental evaluation of the Naive Bayes classifiers implemented in `src/`.

The experiments investigate the effect of vocabulary size, stop-word removal, term-frequency transformations, document-frequency weighting, and Complement Naive Bayes on classification accuracy.

Model performance is evaluated using stratified 10-fold cross-validation. The experiments progressively explore different classifier configurations, beginning with the standard Naive Bayes model and extending to alternative weighting schemes.

## Testing standard Naive Bayes Classifier

```python
import sys
sys.path.insert(0, "..")

from src.data import CSVHandler
from src.evaluation import Evaluator
from src.preprocessing import Preprocessor
from src.classifiers import (
    NaiveBayesClassifier,
    TermTransformedNaiveBayesClassifier,
    DocumentTransformedNaiveBayesClassifier,
    TransformedNaiveBayesClassifier,
    ComplementNaiveBayesClassifier,
    DocumentTransformedComplementNaiveBayesClassifier,
    TransformedComplementNaiveBayesClassifier
)
```

```python
## Load data
csv_handler = CSVHandler('../data/trg.csv', '../data/tst.csv', '../data/predictions.csv')
training_set = csv_handler.read_training_file()
evaluator = Evaluator()

## Test Standard Naive Bayes
n_bayes = NaiveBayesClassifier()
n_bayes_scores = evaluator.cross_validation(n_bayes, training_set, n_words=1000)

mean, std = evaluator.get_stats(n_bayes_scores)
print("Standard Naive Bayes Classifier with 1000 words")
print("mean=", mean)
print("std=", std)
```

## Testing Naive Bayes with stop-word removal
These are words that occur frequently in any document regardless of class, and hence only provide the classifier with noise. Common examples are ‘the’, ‘and’, ‘that’, etc.

```python
n_bayes = NaiveBayesClassifier()
n_bayes_scores5000 = evaluator.cross_validation(n_bayes, training_set, n_words=5000)
print("Now removing stop words...")
n_bayes_scoresU = evaluator.cross_validation(n_bayes, training_set, n_words=5000, remove_useless=True)

mean, std = evaluator.get_stats(n_bayes_scores5000)
print("Standard Naive Bayes Classifier with 5000 words")
print("mean=", mean)
print("std=", std)

mean, std = evaluator.get_stats(n_bayes_scoresU)
print("Standard Naive Bayes Classifier with 5000 words and stop words removed")
print("mean=", mean)
print("std=", std)
```

### Comparing Alternative Classifiers and Weighting Schemes

Having established a baseline, I next investigate whether alternative representations of word frequency improve classification performance.

The following experiments compare logarithmic term-frequency weighting, inverse document-frequency weighting, and their combination. Complement Naive Bayes variants are also evaluated to investigate their effectiveness on this imbalanced dataset.

```python
## Testing Extended Classifiers at higher word counts:
training_set = csv_handler.read_training_file()

tf_bayes = TermTransformedNaiveBayesClassifier()
tf_bayes_scores = evaluator.cross_validation(tf_bayes, training_set, n_words=5000, remove_useless=True)

df_bayes = DocumentTransformedNaiveBayesClassifier()
df_bayes_scores = evaluator.cross_validation(df_bayes, training_set, n_words=5000, remove_useless=True)

tfidf_bayes = TransformedNaiveBayesClassifier()
tfidf_bayes_scores = evaluator.cross_validation(tfidf_bayes, training_set, n_words=5000, remove_useless=True)

## Complement Variants
cn_bayes = ComplementNaiveBayesClassifier()
cn_bayes_scores = evaluator.cross_validation(cn_bayes, training_set, n_words=5000, remove_useless=True)

df_cbayes = DocumentTransformedComplementNaiveBayesClassifier()
df_cbayes_scores = evaluator.cross_validation(df_cbayes, training_set, n_words=5000, remove_useless=True)

tfidf_cbayes = TransformedComplementNaiveBayesClassifier()
tfidf_cbayes_scores = evaluator.cross_validation(tfidf_cbayes, training_set, n_words=5000, remove_useless=True)

## Outputs:
mean, std = evaluator.get_stats(n_bayes_scoresU)
print("Standard Naive Bayes Classifier with 5000 words and stop words removed")
print("mean=", mean)
print("std=", std)

mean, std = evaluator.get_stats(tf_bayes_scores)
print("Term frequency transformed Naive Bayes Classifier with 5000 words")
print("mean=", mean)
print("std=", std)

mean, std = evaluator.get_stats(df_bayes_scores)
print("Document Frequency transformed Naive Bayes Classifier with 5000 words")
print("mean=", mean)
print("std'", std)

mean, std = evaluator.get_stats(tfidf_bayes_scores)
print("Term and Document frequency transformed Naive Bayes Classifier with 5000 words")
print("mean=", mean)
print("std=", std)

mean, std = evaluator.get_stats(cn_bayes_scores)
print("Complement Naive Bayes Classifier with 5000 words")
print("mean=", mean)
print("std=", std)

mean, std = evaluator.get_stats(df_cbayes_scores)
print("Document Frequency Transformed Complement Naive Bayes Classifier with 5000 words")
print("mean=", mean)
print("std=", std)

mean, std = evaluator.get_stats(tfidf_cbayes_scores)
print("Term and Document Frequency Transformed Complement Naive Bayes Classifier with 5000 words")
print("mean=", mean)
print("std=", std)
```

### Increasing Vocabulary Size

The initial experiments suggest that document-frequency weighting and its combination with logarithmic term-frequency weighting offer promising improvements.

I therefore investigate these approaches using a larger vocabulary of 15,000 words, comparing their performance against the standard classifier to determine whether the improvements persist as vocabulary size increases.

```python
## Testing the best of these again with 15,000 words
training_set = csv_handler.read_training_file()

n_bayes = NaiveBayesClassifier() # To ensure improvement is not just from increased word count
n_bayes_scores15000 = evaluator.cross_validation(n_bayes, training_set, n_words=15000, remove_useless=True)

df_bayes = DocumentTransformedNaiveBayesClassifier()
df_bayes_scores = evaluator.cross_validation(df_bayes, training_set, n_words=15000, remove_useless=True)

tfidf_bayes = TransformedNaiveBayesClassifier()
tfidf_bayes_scores = evaluator.cross_validation(tfidf_bayes, training_set, n_words=15000, remove_useless=True)

df_cbayes = DocumentTransformedComplementNaiveBayesClassifier()
df_cbayes_scores = evaluator.cross_validation(df_cbayes, training_set, n_words=15000, remove_useless=True)

mean, std = evaluator.get_stats(n_bayes_scores15000)
print("Standard Naive Bayes Classifier with 15000 words")
print("mean=", mean)
print("std=", std)

mean, std = evaluator.get_stats(df_bayes_scores)
print("Document Frequency transformed Naive Bayes Classifier with 15000 words")
print("mean=", mean)
print("std=", std)

mean, std = evaluator.get_stats(tfidf_bayes_scores)
print("Term and Document frequency transformed Naive Bayes Classifier with 15000 words")
print("mean=", mean)
print("std=", std)

mean, std = evaluator.get_stats(df_cbayes_scores)
print("Document Frequency Transformed Complement Naive Bayes Classifier with 15000 words")
print("mean=", mean)
print("std=", std)

```

Having observed improvements with a vocabulary of 15,000 words, I next investigate whether increasing the vocabulary to 20,000 words provides further improvements in classification accuracy.

```python
tfidf_bayes = TransformedNaiveBayesClassifier()
tfidf_bayes_scores = evaluator.cross_validation(tfidf_bayes, training_set, n_words=20000, remove_useless=True)

mean, std = evaluator.get_stats(tfidf_bayes_scores)
print("Term and Document frequency transformed Naive Bayes Classifier with 20000 words")
print("mean=", mean)
print("std=", std)
```

## Generate predictions for Document-Frequency Complement Naive Bayes Classifier

```python
## Generate predictions for TF-IDF Complement Naive Bayes Classifier
training_set = csv_handler.read_training_file()
test_set = csv_handler.read_testing_file()

preprocessor = Preprocessor()
top_words = preprocessor.set_most_frequent_words(training_set, 20000, remove_useless=True)
processed_training_set = preprocessor.process_training_set(training_set)
processed_test_set = preprocessor.process_test_set(test_set)
print("Completed preprocesing...")

clf = DocumentTransformedComplementNaiveBayesClassifier()
clf.train(processed_training_set, top_words)
print("Completed training...")

evaluator = Evaluator()
print("Training Accuracy for {}".format(clf)) # Essentially informationless, just displayed to ensure its working correctly
print(evaluator.training_accuracy(clf, processed_training_set))

# Retrieve predictions from best performing model
predictions = clf.get_predictions(processed_test_set)
csv_handler.write_output(predictions)
print("Completed predicting...")
```
